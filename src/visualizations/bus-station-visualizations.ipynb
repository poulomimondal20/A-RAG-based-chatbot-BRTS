```python
import pandas as pd

df = pd.read_csv(r'./all_routes_combined.csv')
```

```python
import plotly.express as px

fig = px.scatter_geo(df, lat='Latitude', lon='Longitude', title='Geographical Points', color='Route')
# fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

for route in df['Route'].unique():
    route_df = df[df['Route'] == route]
    
    fig.add_trace(go.Scattergeo(
        lon=route_df['Longitude'],
        lat=route_df['Latitude'],
        mode='lines+markers',
        name=route,
        marker=dict(size=8),
        line=dict(width=2)
    ))

fig.update_layout(
    geo=dict(
        scope='asia',  
        projection_type='natural earth',
        showland=True,
        center=dict(lat=23.24, lon=77.41),  
        projection_scale=250, 
        visible=True
    ),
    margin=dict(l=0, r=0, t=0, b=0),
    xaxis=dict(visible=False),
    yaxis=dict(visible=False),
    autosize=True
)

# fig.show()
```

```python
import folium

color_mapping = {
    'SR1': '#A80A49',  
    'SR2': '#6B8094',  
    'SR3': '#44E221',   
    'SR4': '#75E6F7',
    'SR5': '#C5C344',
    'SR6': '#EB5CF8',
    'SR7': '#FC001E',
    'SR8': '#B16D62',
    'TR1': '#305000',
    'TR2': '#2E8C4A',
    'TR3': '#28180B',
    'TR4': '#9C6AAF',
}

mean_lat = df['Latitude'].mean()
mean_lon = df['Longitude'].mean()

polygon_map = folium.Map(location=[mean_lat, mean_lon], zoom_start=15)

for _, row in df.iterrows():
    fill_color = color_mapping.get(row['Route'], '#00ff00')
    
    folium.RegularPolygonMarker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Station'],
        fill_color=fill_color,
        number_of_sides=25,
        radius=5,
    ).add_to(polygon_map)

polygon_map
```
