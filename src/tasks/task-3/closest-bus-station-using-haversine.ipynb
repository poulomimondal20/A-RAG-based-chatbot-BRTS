### Haversine formula
The Haversine formula is used to calculate the shortest distance between two points on the surface of a sphere (like the Earth), assuming a spherical shape. It provides the great-circle distance, which is the shortest distance over the Earth's surface between two locations, given their latitudes and longitudes.

```python
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0 # radius of the Earth
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    
    distance = R * c
    return distance

def find_closest_bus_station(current_lat, current_lon):
    df['distance'] = df.apply(lambda row: haversine(current_lat, current_lon, row['Latitude'], row['Longitude']), axis=1)
    closest_point = df.loc[df['distance'].idxmin()]
    
    return closest_point
```

```python
sample_current_lat_long = 23.262108, 77.386586
sample_current_lat = sample_current_lat_long[0]
sample_current_lon = sample_current_lat_long[1]

response = find_closest_bus_station(sample_current_lat, sample_current_lon)
```
